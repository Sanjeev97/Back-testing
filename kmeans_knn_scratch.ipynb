````python
# Define distance metric
def euc_dist(value1, value2):
    # value1 and value2 are lists/arrays of features
    distance = 0
    for i in range(len(value1)):
        distance += (value1[i] - value2[i]) ** 2
    return distance ** 0.5

# Identify k nearest neighbors
def k_neighbors(train_data, test_case, k):
    distances = []

    # Calculate distance from test_case to each training point
    for train_row in train_data:
        # Exclude the last element (class label) when calculating distance
        features = train_row[:-1]
        dist = euc_dist(features, test_case)
        distances.append((train_row, dist))

    # Sort by distance and return k nearest neighbors
    distances.sort(key=lambda x: x[1])
    neighbors = [item[0] for item in distances[:k]]

    return neighbors

# Assign class labels (provided - don't modify)
def get_label(train_data, test_case, k):
    neighbors = k_neighbors(train_data, test_case, k)
    labels = [row[-1] for row in neighbors]
    max_label = max(set(labels), key=labels.count)
    return max_label

# Pull it together (provided - don't modify)
def solution(train_data, test_data, k):
    final_labels = list()
    for row in test_data:
        label = get_label(train_data, row, k)
        final_labels.append(label)
    return final_labels
````

````python
import random

def squared_distance(v1, v2):
  return sum((x - y) ** 2 for x, y in zip(v1, v2))

def vector_mean(vectors):
  n = len(vectors)
  return [sum(v[i] for v in vectors) / n
          for i in range(len(vectors[0]))]


class KMeans:
  def __init__(self, k):
      self.k = k # num of clusters
      self.means = None # hold k centroids

  def classify(self, input):
    # return the indes of the cluster closest to the input
    return min(range(self.k),
      key=lambda i: squared_distance(input, self.means[i]))
    pass

  def train(self, inputs):
    # pick k random points as intial centroids
    self.means = random.sample(inputs, self.k)
    assignments = None

    while True:
      # Assign each point to nearest centroid
      new_assignments = [self.classify(p) for p in inputs]

      # Check if assignments have changed, convergence
      if assignments == new_assignments:
        return

      assignments = new_assignments

      # Recompute centroids
      for i in range(self.k):
        # Find all points assigned to cluster i
        i_points = [p for p, a in zip(inputs, assignments) if a == i]

        # update the centroid if cluster isn't empty
        if i_points:
          self.means[i] = vector_mean(i_points)


def squared_clustering_errors(inputs, k):
  # total squared error from k-means clustering
  clusterer = KMeans(k)
  clusterer.train(inputs)
  means = clusterer.means
  assignments = [clusterer.classify(p) for p in inputs]

  return sum(squared_distance(inputs[i], means[cluster])
    for input, cluster in zip(inputs, assignments))


inputs = [
    [-14, -5], [13, 13], [20, 23], [-19, -11], [-9, -16],
    [21, 27], [-49, 15], [26, 13], [-46, 5], [-34, -1],
    [11, 15], [-49, 0], [-22, -16], [19, 28], [-12, -8],
    [-13, -19], [-41, 8], [-11, -6], [-25, -9], [-18, -3]
]


if __name__ == "__main__":

    # Example 1: Three meetups
    print("=== Three Meetups (k=3) ===")
    random.seed(0)
    clusterer = KMeans(3)
    clusterer.train(inputs)
    print(f"Meetup locations: {clusterer.means}")
    print()

    # Show which users go to which meetup
    for i, point in enumerate(inputs):
        cluster = clusterer.classify(point)
        print(f"User at {point} -> Meetup {cluster + 1}")
    print()

    # Example 2: Two meetups
    print("=== Two Meetups (k=2) ===")
    random.seed(0)
    clusterer = KMeans(2)
    clusterer.train(inputs)
    print(f"Meetup locations: {clusterer.means}")
    print()

    # Example 3: Finding the best k using elbow method
    print("=== Elbow Method (finding best k) ===")
    for k in range(1, 6):
        random.seed(0)
        error = squared_clustering_errors(inputs, k)
        print(f"k={k}: total squared error = {error:.1f}")
````
